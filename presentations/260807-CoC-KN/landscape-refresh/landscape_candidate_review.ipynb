## tl;dr

The shortlist prioritizes structural gaps in the landscape rather than raw star rank.

## Context & Methods

- GitHub snapshot: 2026-07-28
- OpenRank: completed months 2026-04 to 2026-06, with material backfill caveats
- Visible WatchEvents: discovery signal only, not exact star growth

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if ROOT.name != 'landscape-refresh':
    ROOT = Path('presentations/260807-CoC-KN/landscape-refresh')
shortlist = pd.read_csv(ROOT / 'data/human_review_shortlist.csv')
quality = pd.read_csv(ROOT / 'data/data_quality_checks.csv')
shortlist.shape, quality.shape
```

## Data

```python
shortlist[['decision','repo_name','target_layer','target_box','stars_current','openrank_202604','openrank_202605','openrank_202606']].head(20)
```

## Results

```python
a = shortlist[shortlist.decision.str.startswith('A-')].sort_values('stars_current')
colors = ['#3777B7' if x == 'Agent Infra' else '#B87819' for x in a.target_layer]
ax = a.plot.barh(x='repo_name', y='stars_current', color=colors, edgecolor='#27313A', legend=False, figsize=(11, 7))
ax.set_title('A-tier candidate GitHub stars snapshot — 2026-07-28')
ax.set_xlabel('GitHub stars (attention snapshot, not community health)')
ax.set_ylabel('')
ax.grid(axis='x', color='#D9DEE3', linewidth=.7)
plt.tight_layout();
```

## Takeaways

1. The strongest omissions are context, protocol, sandbox orchestration, and state reuse.
2. Current stars are useful for attention, not maturity.
3. The Large Models layer needs a separate model-release and openness review.
